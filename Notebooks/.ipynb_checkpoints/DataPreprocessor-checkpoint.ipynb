```python
import pandas as pd
import logging

# Configure logging for this module
logger = logging.getLogger(__name__)


class DataProcessor:
    """
    A production-ready DataFrame processing utility class.
    Provides methods for column normalization, filtering, dropping NaNs, and merging.
    """

    def modify_columns(self, df: pd.DataFrame) -> pd.DataFrame:
        """
        Normalize column names: strip spaces, lowercase, replace spaces with underscores.

        Args:
            df (pd.DataFrame): Input DataFrame.

        Returns:
            pd.DataFrame: DataFrame with normalized column names.
        """
        try:
            df = df.copy()
            df.columns = [col.strip().lower().replace(' ', '_')
                          for col in df.columns]
            logger.info("Columns normalized successfully.")
            return df
        except Exception as e:
            logger.exception(f"Error in modify_columns: {e}")
            return df

    def filter_df(self, df: pd.DataFrame, filters: dict = None) -> pd.DataFrame:
        """
        Filter a DataFrame based on multiple column equality checks.
        Automatically handles spaces and capitalization in column names.
        Missing columns are ignored.

        Args:
            df (pd.DataFrame): Input DataFrame.
            filters (dict, optional): Dictionary of column:value pairs for filtering.

        Returns:
            pd.DataFrame: Filtered DataFrame.
        """
        try:
            filtered_df = df.copy()
            if filters:
                norm_filters = {col.strip().lower().replace(
                    ' ', '_'): val for col, val in filters.items()}
                for col, val in norm_filters.items():
                    if col not in filtered_df.columns:
                        logger.warning(
                            f"Filter column '{col}' not found; skipping filter.")
                        continue
                    filtered_df = filtered_df[filtered_df[col] == val]
            return filtered_df
        except Exception as e:
            logger.exception(f"Error in filter_df: {e}")
            return df

    def dropna_columns(self, df: pd.DataFrame, non_null_cols: list) -> pd.DataFrame:
        """
        Drop rows where any of the specified columns are NaN.

        Args:
            df (pd.DataFrame): Input DataFrame.
            non_null_cols (list): List of column names that must not be null.

        Returns:
            pd.DataFrame: DataFrame with specified columns non-null.
        """
        try:
            if not non_null_cols:
                return df
            norm_non_null_cols = [col.strip().lower().replace(' ', '_')
                                  for col in non_null_cols]
            for col in norm_non_null_cols:
                if col not in df.columns:
                    logger.warning(
                        f"Non-null column '{col}' not found; skipping dropna for this column.")
                    norm_non_null_cols.remove(col)
            if norm_non_null_cols:
                df = df.dropna(subset=norm_non_null_cols)
            return df
        except Exception as e:
            logger.exception(f"Error in dropna_columns: {e}")
            return df

    def merge_df(self, df1: pd.DataFrame, df2: pd.DataFrame, on: str, how: str = "inner",
                 suffixes=('_df1', '_df2')) -> pd.DataFrame:
        """
        Merge two DataFrames on a specified column, after normalizing the column.

        Args:
            df1 (pd.DataFrame): Left DataFrame.
            df2 (pd.DataFrame): Right DataFrame.
            on (str): Column name to merge on.
            how (str): Type of merge ('inner', 'left', 'right', 'outer').
            suffixes (tuple): Suffixes to apply to overlapping columns.

        Returns:
            pd.DataFrame: Merged DataFrame.
        """
        try:
            on_norm = on.strip().lower().replace(' ', '_')
            if on_norm not in df1.columns or on_norm not in df2.columns:
                logger.warning(
                    f"Merge column '{on}' not found in one of the DataFrames; returning df1.")
                return df1
            merged_df = pd.merge(df1, df2, on=on_norm,
                                 how=how, suffixes=suffixes)
            logger.info("DataFrames merged successfully.")
            return merged_df
        except Exception as e:
            logger.exception(f"Error in merge_df: {e}")
            return df1

    def get_processed_data(self, df: pd.DataFrame, filters: dict = None, non_null_cols: list = None) -> pd.DataFrame:
        """
        Normalize columns, optionally filter data, and optionally drop NaNs.

        Args:
            df (pd.DataFrame): Input DataFrame.
            filters (dict, optional): Filter conditions.
            non_null_cols (list, optional): Columns to enforce non-null.

        Returns:
            pd.DataFrame: Processed DataFrame.
        """
        try:
            df = self.modify_columns(df)
            if filters:
                df = self.filter_df(df, filters=filters)
            if non_null_cols:
                df = self.dropna_columns(df, non_null_cols=non_null_cols)
            return df
        except Exception as e:
            logger.exception(f"Error in get_processed_data: {e}")
            return df

    def get_merged_data(self, df1: pd.DataFrame, df2: pd.DataFrame, on: str, how: str = "inner",
                        suffixes=('_df1', '_df2'), filters_merged: dict = None,
                        non_null_cols_merged: list = None) -> pd.DataFrame:
        """
        Merge two DataFrames and optionally filter and drop NaNs after the merge.

        Args:
            df1 (pd.DataFrame): Left DataFrame.
            df2 (pd.DataFrame): Right DataFrame.
            on (str): Column name to merge on.
            how (str): Type of merge ('inner', 'left', 'right', 'outer').
            suffixes (tuple): Suffixes to apply to overlapping columns.
            filters_merged (dict, optional): Filter conditions applied after merge.
            non_null_cols_merged (list, optional): Columns to enforce non-null after merge.

        Returns:
            pd.DataFrame: Merged and processed DataFrame.
        """
        try:
            df1 = self.modify_columns(df1)
            df2 = self.modify_columns(df2)

            merged_df = self.merge_df(
                df1, df2, on=on, how=how, suffixes=suffixes)

            if filters_merged:
                merged_df = self.filter_df(merged_df, filters=filters_merged)
            if non_null_cols_merged:
                merged_df = self.dropna_columns(
                    merged_df, non_null_cols=non_null_cols_merged)

            return merged_df
        except Exception as e:
            logger.exception(f"Error in get_merged_data: {e}")
            return merged_df
```
